```python
import pandas as pd
```

```python
df= pd.read_csv('../datasets/processed/df_15min.csv')
df.head()
```

```python
print(df[["SCATS Number", "NB_LATITUDE", "NB_LONGITUDE"]].drop_duplicates().head(20))
```

```python
df_update = df[["SCATS Number", "NB_LATITUDE", "NB_LONGITUDE"]].drop_duplicates()
df_update
```

```python
df_cleaned = df_update.groupby("SCATS Number")[["NB_LATITUDE", "NB_LONGITUDE"]].mean().reset_index()
df_cleaned
```

```python
df_cleaned.to_csv("../datasets/processed/SCAT_cord.csv", index=False)
```
