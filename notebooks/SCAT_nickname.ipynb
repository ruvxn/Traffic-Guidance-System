```python
import pandas as pd
```

```python
df = pd.read_csv("../datasets/processed/Scats_Data_October_2006_cleaned.csv")
df.head
```

```python
unique_scats = df[["SCATS Number", "Main Road", "Cross Road"]].drop_duplicates()
```

```python
def format_road_name(name):
    
    name = name.replace("_RD", "").replace("_ST", "").replace("_AV", "").replace("_HWY", "").replace("_FWY", "")
    return ''.join(word.capitalize() for word in name.split())
```

```python
unique_scats["Nickname"] = unique_scats.apply(
    lambda row: format_road_name(row["Main Road"]) + format_road_name(row["Cross Road"]), axis=1
)
```

```python
nickname_mapping = unique_scats[["SCATS Number", "Nickname"]].drop_duplicates(subset="SCATS Number")
```

```python
nickname_mapping.to_csv("../datasets/processed/SCATS_Nicknames.csv", index=False)
```
